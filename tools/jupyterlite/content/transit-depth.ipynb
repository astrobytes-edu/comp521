# Transit depth experiment
Change `p`, run the cell, then compare the result with your prediction.

```python
import matplotlib.pyplot as plt
p = 0.10
depth = p**2
print(f'blocked fraction = {depth:.3f}')
print(f'normalized flux = {1-depth:.3f}')
values = [i / 100 for i in range(2, 51)]
plt.plot(values, [value**2 for value in values])
plt.scatter([p], [depth])
plt.xlabel('radius ratio p')
plt.ylabel('blocked fraction')
plt.show()
```
