# Chapter 6 micro-experiment: a temperature invariant

**Predict.** If `Temperature(-1)` is constructed, should it succeed or raise `ValueError`? Write a one-sentence reason before running the next cell.

```python
class Temperature:
    """Temperature with a non-negative Kelvin invariant."""

    def __init__(self, kelvin):
        self.kelvin = kelvin

    @property
    def kelvin(self):
        return self._kelvin

    @kelvin.setter
    def kelvin(self, value):
        if value < 0:
            raise ValueError('Below absolute zero!')
        self._kelvin = value

    @property
    def celsius(self):
        return self.kelvin - 273.15
```

## Run the prediction

The `try`/`except` keeps the intentional invalid input visible without stopping the notebook.

```python
try:
    Temperature(-1)
except ValueError as error:
    print(type(error).__name__, error)

absolute_zero = Temperature(0)
print(f'{absolute_zero.kelvin:.0f} K = {absolute_zero.celsius:.2f} degC')
```
