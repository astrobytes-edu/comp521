Synthetic data for presentation testing; not an observed source.

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({"font.size": 16})
t = np.linspace(0, 4, 81)
flux = 1 + 0.1 * np.cos(2 * np.pi * t)
fig, ax = plt.subplots(figsize=(8, 3.5))
ax.plot(t, flux, color="#2f6f6f", linewidth=2)
ax.set(xlabel="Time (arbitrary units)", ylabel="Relative flux")
fig.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-2, 2, 81)
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, x**2, color="#2f6f6f")
ax.set(xlabel="x", ylabel="y = x²")
fig.tight_layout(); plt.show()
```
