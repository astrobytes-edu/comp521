# Losses, gradients, and safeguarded optimization: executed calculations

All examples are synthetic. The fading model is dimensionless excess-flux relaxation: $f(t)=A\exp(-t/\tau)$, with $t=\mathrm{time}/t_0$ and $y=F/F_0$. It is not a full stellar model. Frozen parameters: $A=1.8$, $\tau=2.4$, independent Gaussian $\sigma=0.08$, 16 observation times from 0 to 6, and NumPy seed 20261110.

The scalar quadratic and anisotropic bowl have analytic solutions. The inspectable LM implementation solves augmented least squares, accepts actual loss decreases, and is checked against SciPy. The notebook uses the same functions that generate the slide assets. These checks establish behavior of teaching examples, not scientific validation.

```python
import os, sys
sys.dont_write_bytecode = True
os.environ["MPLCONFIGDIR"] = "/private/tmp/comp521-optimization-mpl"
from pathlib import Path
assert Path(sys.executable).parent.parent.name == ".venv", "Use the course .venv kernel."
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
import figures
from figures import residual, jacobian, loss, fit_lm, START
receipt = figures.numerical_checks(write=True)
print("Course kernel; float64; fixed seed", receipt["seed"])
print("LM parameters A,tau:", receipt["lm"]["fit_parameters"])
print("LM attempts/accepted/rejected:", receipt["lm"]["attempts"], receipt["lm"]["accepted"], receipt["lm"]["rejected"])
print("Weighted least-squares loss:", receipt["lm"]["loss"])
print("SciPy log-parameter difference:", receipt["lm"]["max_log_parameter_difference_from_scipy"])
print("All focused checks:", receipt["status"])
```

## One executed proposal

This is the first proposal from the initial $A=0.7$, $\tau=0.7$ and $\nu=1$. It is evaluated, not automatically accepted. The full teaching solver increases damping after rejection.

```python
theta = START.copy()
q, J = residual(theta), jacobian(theta)
nu = 1.0
augmented_J = np.vstack([J, np.sqrt(nu) * np.eye(2)])
augmented_rhs = np.r_[-q, np.zeros(2)]
step = np.linalg.lstsq(augmented_J, augmented_rhs, rcond=None)[0]
trial = theta + step
print("Current loss:", float(loss(theta)))
print("Trial loss:", float(loss(trial)))
print("Accept:", bool(loss(trial) < loss(theta)))
```
