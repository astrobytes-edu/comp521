# Probability and uncertainty: executed figure calculations

Other distributions, linear maps, and thresholds are explicitly illustrative. All figures are generated by the same `figures.py` functions that write the slide assets. NumPy uses float64; the Monte Carlo seed is fixed and estimates are compared with analytic answers in units of their standard errors.
